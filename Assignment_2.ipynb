```python
##Question 1
rollNo = "1024170205"
L = [int(digit) * 10 for digit in rollNo]
print("Original L:", L)
L.append(203)
print("After append(203):", L)
L.insert(3, 155)
print("After insert(3, 155):", L)
L.remove(155)
print("After remove(155):", L)
L.pop()
print("After pop():", L)
L.sort()
print("Sorted ascending:", L)
L.sort(reverse = True)
print("Sorted descending:", L)
print("First three elements:", L[:3])
print("Last three elements:", L[-3:])
avg = sum(L) / len(L)
above_avg = [x for x in L if x > avg]
print("Average of L:", avg)
print("Elements greater than average:", above_avg)
```

```python
##Question 2
scores = tuple(L[:8])
print("scores tuple:", scores)
highest = max(scores)
highest_index = scores.index(highest)
lowest = min(scores)
lowest_count = scores.count(lowest)
print(f"Highest score: {highest} at index {highest_index}")
print(f"Lowest score: {lowest}, appears {lowest_count} time(s)")
reversed_scores = list(scores[::-1])
print("Reversed scores (as list):", reversed_scores)
user_score = int(input("Enter a score to search for: "))
if user_score in scores:
    print(f"{user_score} found at index {scores.index(user_score)}")
else:
    print(f"{user_score} not present in scores")
try:
    scores[0] = 100
except TypeError as e:
    print("Error caught:", e)
first_score, second_score, *remaining_scores = scores
print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

```python
##Question 3
import random
from collections import Counter
rollNo = 1024170205
random.seed(rollNo)
numbers = [random.randint(100, 900) for _ in range(100)]
print("Generated numbers:", numbers)
odd_numbers = [n for n in numbers if n % 2 != 0]
print("Odd numbers count:", len(odd_numbers))
even_numbers = [n for n in numbers if n % 2 == 0]
print("Even numbers count:", len(even_numbers))
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

primes = [n for n in numbers if is_prime(n)]
print("Prime numbers count:", len(primes))
print("Prime numbers list:", primes)
counts = Counter(numbers)
most_common_num, freq = counts.most_common(1)[0]
print(f"Most frequent number: {most_common_num} (occurs {freq} times)")
```

```python
##Question 4
digits = [1, 0, 2, 4, 1, 7, 0, 2]
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}
print('Set A : ', A)
print('Set B : ', B)
union_AB  = A | B
print('Union(A, B) : ', union_AB)
intersection_AB = A & B
print('Intersection (A, B) : ', intersection_AB)
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)
print('A - B : ', A_minus_B)
print('B - A : ', B_minus_A)
print("Note: difference() gives elements in one set but not the other (one direction at a time),")
print("while symmetric_difference() gives elements in either set but NOT in both (combines both directions).")
sym_diff = A.symmetric_difference(B)
print('Symmetric difference(A, B) : ', sym_diff)
is_subset = A.issubset(B)
is_superset = B.issuperset(A)
print("Is A a subset of B?", is_subset)
print("Is B a superset of A?", is_superset)
X = int(input("Enter a value X to remove from set A: "))
A.discard(X)
print("Set A after discard():", A)
print("discard() is safer than remove() because discard() does nothing if X is not in the set,")
print("while remove() raises a KeyError, potentially crashing the program.")
```
