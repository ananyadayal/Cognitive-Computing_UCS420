```python
##Question 1
import numpy as np
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
##1)a)
corrected = temperature + 2
print("a. Corrected temperatures:", corrected)
##1)b)
fahrenheit = (9 / 5) * corrected + 32
print("b. Fahrenheit:", fahrenheit)
##1)c)
mask = corrected > 32
print("c. Readings > 32 C:", corrected[mask])
##1)d)
print("d. Count > 32 C:", np.sum(mask))

##1)e)
print("""When we use a for loop, Python goes through the readings one by one
and does the work for each value separately, which is slow when there is a lot
of data. With vectorization, we write something like temperature + 2 and NumPy
applies it to the whole array in one go. It does this in fast, pre-compiled code
behind the scenes, so it is quicker and the code is much shorter. Boolean indexing
works the same way. Instead of looping and checking if temp > 32 for every value,
we make one True/False mask and use it to pull out all the matching readings at
once. Counting them is also easy, because np.sum(mask) counts the True values.
Together they make the code faster, cleaner and easier to read.""")
```

```python
##Question 2
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]])
##2)a)
print("Total steps:", steps.sum())
##2)b)
print("Mean steps:", steps.mean())
##2)c)
print("Max:", steps.max(), "| Min:", steps.min())
##2)d)
print("Total per day (axis=0):", steps.sum(axis=0))
##2)e)
print("Total per user (axis=1):", steps.sum(axis=1))
##2)f)
pos = tuple(int(i) for i in np.unravel_index(steps.argmax(), steps.shape))
print("argmax (flat index):", steps.argmax())
print("Position (user, day) :", pos,
      "-> User", pos[0] + 1, ", Day", pos[1] + 1,
      "with", steps[pos], "steps")
```

```python
##Question 3
##3)a)
original = np.array([1, 2, 3, 4, 5, 6])
##3)b)
subset = original[1:5]
##3)c)
subset[0] = 999
print("original:", original)
print("   subset  :", subset, "(original changed, because a slice is a view)")

##3)d)
original = np.array([1, 2, 3, 4, 5, 6])
copied = original[1:5].copy()
copied[0] = 500
print("original:", original)
print("   copied  :", copied, "(original unchanged)")

##3)e)
matrix = np.arange(1, 13).reshape(3, 4)
print("Matrix:\n", matrix)

##3)f)
print("First row     :", matrix[0])
print("Last row      :", matrix[-1])
print("Second column :", matrix[:, 1])
print("Rows 1-2, cols 2-3:\n", matrix[0:2, 1:3])

##3)g)
flat = matrix.flatten()
rav = matrix.ravel()
print("flatten():", flat)
print("ravel()  :", rav)

##3)h)
rav[0] = 100
print("After rav[0] = 100, matrix[0, 0] =", matrix[0, 0],
      "-> original affected (ravel returns a view when possible)")

##3)i)
flat[1] = 200
print("After flat[1] = 200, matrix[0, 1] =", matrix[0, 1],
      "-> original NOT affected (flatten always returns a copy)")

##3)j)
print("shape:", matrix.shape, "| ndim:", matrix.ndim,
      "| size:", matrix.size, "| dtype:", matrix.dtype)
```

```python
##Question 4
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]])
y = np.array([40, 65, 30, 85])

##4)a)
print("X shape:", X.shape, "| ndim:", X.ndim)

##4)b)
print("X.T:\n", X.T)
print("""The transpose X.T flips the matrix so that rows become columns and
columns become rows. X is 4×3, with one row per user and one column per feature,
so X.T is 3×4. Each row of X.T now holds one feature across all four users. This
layout is needed for the OLS formula, because multiplying X.T @ X combines the
features with each other and gives a square 3×3 matrix we can invert.""")

##4)c)
XtX = X.T @ X
print("X.T @ X:\n", XtX)

##4)d)
print("det(X.T @ X) =", round(np.linalg.det(XtX), 2), "(non-zero, so the inverse exists)")
XtX_inv = np.linalg.inv(XtX)
print("inv(X.T @ X):\n", XtX_inv)

##4)e)
beta = XtX_inv @ X.T @ y
print("beta (sleep, activity, stress):", beta)
print("Check with np.linalg.lstsq:", np.linalg.lstsq(X, y, rcond=None)[0])

##4)f)
print("""The coefficients tell us how much each feature affects the predicted assistance score.
Each one is the change in the score for a one-unit increase in that feature, when the other two
stay the same. We got about -0.16 for sleep hours, 0.14 for activity level and 10.06 for stress
level. Stress has by far the biggest effect, so each extra point of stress raises the predicted
assistance score by about 10, meaning more stressed users are predicted to need more help.Activity
has a small positive effect, and sleep has a very small negative one. The sleep sign is a bit
unexpected, but we only have 4 users and no intercept term, so these values are best treated as
a practice example and not a reliable model.""")

##4)g)
new_user = np.array([5, 40, 7])
prediction = new_user @ beta
print("Predicted assistance score for", new_user, "=", round(prediction, 2))
```
